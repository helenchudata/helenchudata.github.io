# 3. Services Dataset Wrangling

## <span style="color:red;">3A. Standardize Column Titles</span>

**The code cell below is graded. Do not delete the cell.**

```python
# WRITE YOUR SOLUTION HERE. DO NOT DELETE. THIS CELL IS GRADED.
services_df.columns = services_df.columns.str.strip().str.lower().str.replace(' ', '_')
```

## <span style="color:red;">3B. Convert Data Types</span>


**The code cell below is graded. Do not delete the cell.**

```python
# WRITE YOUR SOLUTION HERE. DO NOT DELETE. THIS CELL IS GRADED.
services_df['customer_enrollment_date'] = pd.to_datetime(services_df['customer_enrollment_date'])
```

# 4. Merge the Dataframes

## 4A. Identify the Connecting Columns

```python
print(demog_df['customer_id'].unique())
```

```python
print(services_df['customer_id'].unique())
```

## <p style="color:red;">4B. Manipulate the Connecting Columns</p>

**The code cell below is graded. Do not delete the cell.**

```python
# WRITE YOUR SOLUTION HERE. DO NOT DELETE. THIS CELL IS GRADED.
demog_df['customer_id'] = demog_df['customer_id'].astype(str).str.strip()
services_df['customer_id'] = services_df['customer_id'].astype(str).str.strip()
```

## <p style="color:red;">4C. Merge the DataFrames</p>
**The code cell below is graded. Do not delete the cell.**

```python
# WRITE YOUR SOLUTION HERE. DO NOT DELETE. THIS CELL IS GRADED.
tco_df = pd.merge(demog_df, services_df, on='customer_id')
```

## <p style="color:red;">5B. Calculate customer tenure in days</p>
**The code cell below is graded. Do not delete the cell.**

```python
# WRITE YOUR SOLUTION HERE. DO NOT DELETE. THIS CELL IS GRADED.
today_date = pd.to_datetime('2020-09-01')

tco_df['customer_tenure'] = (today_date - tco_df['customer_enrollment_date']).dt.days

average_tenure = tco_df['customer_tenure'].mean()

print(f"Average customer tenure: {average_tenure} days")
```
